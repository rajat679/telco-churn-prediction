```python
import pandas as pd

demographics = pd.read_csv("../data/enrichment/Telco_customer_churn_demographics.csv")
location = pd.read_csv("../data/enrichment/Telco_customer_churn_location.csv")
services = pd.read_csv("../data/enrichment/Telco_customer_churn_services.csv")
status = pd.read_csv("../data/enrichment/Telco_customer_churn_status.csv")
population = pd.read_csv("../data/enrichment/Telco_customer_churn_population.csv")

for name, d in [("demographics", demographics), ("location", location),
                ("services", services), ("status", status), ("population", population)]:
    print(name, "-> shape:", d.shape)
    print(d.columns.tolist())
    print()
```

```python
# Cell — load main dataset fresh in this notebook
df = pd.read_csv("../data/raw/CustomerChurn.csv")

# Fix Total Charges again (this notebook is separate, so redo the Week 2 fix)
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce').fillna(0)

print("Main df shape before merge:", df.shape)
```

```python
# Cell — merge demographics, keeping only new columns
demo_cols = demographics[['Customer ID', 'Age', 'Married', 'Number of Dependents']]

df = df.merge(demo_cols, on='Customer ID', how='left')
print("Shape after demographics merge:", df.shape)
df[['Customer ID', 'Age', 'Married', 'Number of Dependents']].head()
```

```python
print("Shape after demographics merge:", df.shape)
```

```python
# Cell — merge status, keeping only new/valuable columns
status_cols = status[['Customer ID', 'Satisfaction Score', 'CLTV', 'Churn Category', 'Churn Reason']]

df = df.merge(status_cols, on='Customer ID', how='left')
print("Shape after status merge:", df.shape)
df[['Customer ID', 'Satisfaction Score', 'CLTV', 'Churn Category', 'Churn Reason']].head()
```

```python
# Cell — merge location, keeping only Zip Code
location_cols = location[['Customer ID', 'Zip Code']]

df = df.merge(location_cols, on='Customer ID', how='left')
print("Shape after location merge:", df.shape)
df[['Customer ID', 'Zip Code']].head()
```

```python
# Cell — merge population via Zip Code
pop_cols = population[['Zip Code', 'Population']]

df = df.merge(pop_cols, on='Zip Code', how='left')
print("Shape after population merge:", df.shape)
df[['Customer ID', 'Zip Code', 'Population']].head()
```

```python
# Cell — final structure check
print(df.shape)
df.info()
```

```python
# Cell — check for any unexpected new nulls introduced by merging
df.isnull().sum()
```

```python
# Cell — save the merged, cleaned dataset
df.to_csv("../data/processed_customer_churn.csv", index=False)
print("Saved:", df.shape)
```

```python
# Cell — fix Population dtype
df['Population'] = df['Population'].str.replace(',', '').astype(int)
print(df['Population'].dtype)
df['Population'].describe()
```
